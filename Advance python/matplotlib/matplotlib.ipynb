## ploting line

```python
import matplotlib.pyplot as plt
x=['Mon','Tue','Wed','Thus','Fri'] # x-axis
y=[10,15,7,20,12] # y-axis
plt.plot(x,y)
plt.title('Bakery Sales This Week!')
plt.xlabel('Day of the week')
plt.ylabel('Sales Per Day')
plt.show()
```

```python
import matplotlib.pyplot as plt
months=[1,2,3,4]  # x-axis
sales=[1000,1500,1200,1800]  # y-axis
plt.plot(months,sales,
         color='blue',
         linestyle='--',
         linewidth=2,
         marker='o',
         label='2025 sales data')
plt.xlabel('Months')
plt.ylabel('Sales per month')

# plt.title()
plt.title('Monthly Sales Data Report')
# plt.legend(location='upper left' or lower right,fontsize=12)
plt.legend(loc='lower right',fontsize=10)
plt.grid(color='gray',linestyle=':',linewidth=1)
plt.xlim(1,4)
plt.ylim(0,2000)
plt.xticks([1,2,3,4],['M1','M2','M3','M4'])

plt.show()
```

## Pie Chart

```python

# plt.pie(values,labels=label_list,color=color_list,autopct='%1.1f%%')
```

```python
import matplotlib.pyplot as plt

regions = ['North', 'South', 'East', 'West']
revenue = [3000, 2000, 1500, 1000]

# १. सेतो क्यान्भास/फ्रेमलाई ठूलो बनाउने (८ इन्च चौडाइ, ६ इन्च उचाइ)
plt.figure(figsize=(8, 6))

plt.pie(
    revenue,
    labels=regions,
    autopct='%1.1f%%',
    colors=['gold', 'skyblue', 'lightgreen', 'coral'],
)

plt.title('Revenue Contribution by Region')

# २. Legend लाई Pie Chart भन्दा बाहिर (Right side मा) सार्ने
plt.legend(title="Regions", loc="center left", bbox_to_anchor=(1.2, 0.5))

# ३. ग्राफका तत्वहरू आपसमा नथिचियोस् भनेर लेआउट मिलाउने
plt.tight_layout()

plt.show()
```

```python
# plt.scatter(x,y,color='color_name',marker='marker',label='label_name')
```

```python
import matplotlib.pyplot as plt
hours_studies=[1,2,3,4,5,6,7,8]
exam_scores=[50,55,60,65,70,75,80,85]
# scattre plot
plt.scatter(hours_studies,exam_scores,color='green',marker='^',label='Student Data')
plt.xlabel('Hours studied')
plt.ylabel('Exam Score')
plt.title('Relationship Between Study Time and Exam Score')
plt.legend(loc='upper left')
plt.grid(True)
plt.show()
```

```python
import matplotlib.pyplot as plt
plt.scatter([1,2,3],[50,60,70],color='blue',label='Class A') # g1
plt.scatter([1,2,3],[45,50,52],color='orange',label='Class B') # g2
# scattre plot
plt.xlabel('Hours studied')
plt.ylabel('Exam Score')
plt.title('Comparision of Two Classes')
plt.legend(loc='upper left')
plt.grid(True)
plt.show()
```

```python
import matplotlib.pyplot as plt
x=[1,2,3,4]
y=[10,20,15,25]

plt.subplot(1,2,1)  # 1 row ,2 column  1st subplot
plt.title('Line chart')
plt.plot(x,y)

plt.subplot(1,2,2)   # 1 row ,2 column, 2nd subplot
plt.bar(x,y)

plt.title('Bar Chart')

plt.show()
```

```python
import matplotlib.pyplot as plt
x=[1,2,3,4]
y=[10,20,15,25]

# fig,ax=plt.subplot(nrows,ncols,figsize=(width,height))
fig,ax=plt.subplots(1,2,figsize=(10,5))
ax[0].plot(x,y)
ax[0].set_title('Line Plot')
ax[1].bar(x,y,color='green')
ax[1].set_title('Bar Chart')\

fig.suptitle('Comparison of Line Plot and Bar Chart')
plt.tight_layout()
plt.show()
```
