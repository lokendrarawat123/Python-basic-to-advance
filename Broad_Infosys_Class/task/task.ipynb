### Case Scenario 1: E-Commerce Dynamic Shipping Engine
#### • Core Concept: Control Structures (if-elif-else and nested conditions)
#### • Difficulty: Basic
#### • Scenario: An online store calculates delivery charges dynamically. Orders with a cart 
#### value over $150 receive free shipping. If the cart value is under $150, look up the 
#### destination region: "Domestic" shipping costs a flat fee of $10, while "International" 
#### shipping costs $25. However, if the customer holds a premium membership pass, they 
#### receive a $5 discount on any calculated shipping fee under $150, but the final shipping cost 
#### can never drop below $0.
#### • Task: Code a program that asks the user to input their cart total, geographic delivery zone, 
#### and premium member status, then calculates and prints the precise final shipping cost. 

```python
cart = int(input('Enter your cart amount: '))
shipping_cost = 0

if cart < 0:
    print('Invalid cart amount')

elif cart > 150:
    print('Your shipping cost is free ($0). Your order will be delivered.')
    shipping_cost = 0
    print(f'Your final shipping cost is = ${shipping_cost}')

else:
    location = input("Enter delivery zone (Domestic/International): ")

    if location.lower() == 'domestic':
        print('Your shipping cost is $10')
        shipping_cost = 10

    elif location.lower() == 'international':
        print('Your shipping cost is $25')
        shipping_cost = 25

    else:
        print('Invalid location')
        shipping_cost = None

    if shipping_cost:
        membership = input("Are you a premium member? (yes/no): ")

        if membership.lower() == 'yes':
            if shipping_cost >= 5:
                shipping_cost -= 5
            else:
                shipping_cost = 0

        elif membership.lower() == 'no':
            pass

        else:
            print("Invalid membership status!")

        print(f'Your final shipping cost is = ${shipping_cost}')
```
