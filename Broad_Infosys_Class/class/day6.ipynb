### While loop

```python
'''
while condition:
    logic...
--- program runs until condtion becomes False.
'''
```

```python
# Number guess
secret_number=7
guess_number=0
while guess_number!=secret_number:
    guess_number=int(input('Enter a number: '))
    if guess_number==secret_number:
        print('guess number is Correct.')
    else:
            print('guess number is incorrect.Try again.')
```

```python
#  Using without 
secret_number=7
# data=True
# while data:
while True:
    guess_number=int(input('Enter a number: '))
    if guess_number==secret_number:
        print('guess number is Correct.')
        # date=False
        break
    else:
            print('guess number is incorrect.Try again.')
```

```python
secret_number=7
count=3

while True:
    guess_number=int(input('Enter a number: '))
    print(f'Remaing attempts is {count}')
    if guess_number==secret_number:
        print('guess number is Correct.')
        break
       
    else:
        print('guess number is incorrect.Try again.')
        count-=1
    if count==0:
        print('No attempt left')
        break
   
```

### enumerate


```python
'''
-Loops both index and value of list.
'''
```

```python
alpha=['a','b','c','d','e','f']
for idx,val in enumerate(alpha):
    print(f'{idx}:{val}')
```

```python
multi_d=[[1,2,3],[4,5,6],[7,8,9]]
#1st loop -->row
# 2nd loop-->column
for row in multi_d:
    for col in row:
        print(col)
```

```python
for r_idx,row in enumerate(multi_d):
    print(f'{r_idx}:{row}')
    for c_idx,col in enumerate(row):
        print(f'{c_idx}:{col}')
    print('\n')
        
```

### slicing


```python
# list[start:stop:step]
alpha=['a','b','c','d','e','f','g','i','j']
alpha[1:6]
alpha[:6]
alpha[3:]
alpha[1:6:2]
alpha[::2]
alpha[::-1] # for the reverse list.
```
