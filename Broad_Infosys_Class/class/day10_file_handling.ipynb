## Dictionary Data to CSV file

```python
products=[
    {
        'Product Name':'Laptop',
        'Product Brand':'HP',
        'Product Price':'$5000',
    },
        {
        'Product Name':'Charger',
        'Product Brand':'NavyForce',
        'Product Price':'$2000',
    },
    {
        'Product Name':'Watch',
        'Product Brand':'anker',
        'Product Price':'$299.99',
    }]
```

```python
with open('products.csv','w',newline='') as file:
    columns=['Product Name','Product Brand','Product Price']
    writer=csv.DictWriter(file,fieldnames=columns)
    writer.writeheader()
    writer.writerows(products)
```

```python
## Read file
```

```python
with open('products.csv','r') as file:
    reader=csv.DictReader(file)
    for row in reader:
        for key,val in row.items():
            print(f'{key}:{val}')
        print('\n')
                         
```

## Json (java script Object Notaion) File


```python
# dump -> Create
# Load -> Read
```

```python
products_json=[
    {
        'Product Name':'Laptop',
        'Product Brand':'HP',
        'Product Price':'$5000',
    },
        {
        'Product Name':'Charger',
        'Product Brand':'NavyForce',
        'Product Price':'$2000',
    },
    {
        'Product Name':'Watch',
        'Product Brand':'anker',
        'Product Price':'$299.99',
    }]
```

### create Json File

```python
import json
```

```python
# open(filename,operation)
with open('products.josn','w') as file:
    json.dump(products_json,file,indent=4)
```

```python
# read(filename,operation)
with open('products.josn','r') as file:
    data=json.load(file)
    for row in data:
        for idx,val in row.items():
            print(f'{idx}:{val}')
        print('\n')
```

### Append data into json

```python
new_product_json={
        'Product Name':'mobile',
        'Product Brand':'samsung',
        'Product Price':'$699.99',
    }
```

```python
### each time replace when we append the new json data.
with open('products.json','w') as file:
    data.append(new_product_json)
    json.dump(data,file)
```
