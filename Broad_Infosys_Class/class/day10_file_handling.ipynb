## Dictionary Data to CSV file

```python
products=[
    {
        'Product Name':'Laptop',
        'Product Brand':'HP',
        'Product Price':'$5000',
    },
        {
        'Product Name':'Charger',
        'Product Brand':'NavyForce',
        'Product Price':'$2000',
    },
    {
        'Product Name':'Watch',
        'Product Brand':'anker',
        'Product Price':'$299.99',
    }]
```

```python
with open('products.csv','w',newline='') as file:
    columns=['Product Name','Product Brand','Product Price']
    writer=csv.DictWriter(file,fieldnames=columns)
    writer.writeheader()
    writer.writerows(products)
```

```python
## Read file
```

```python
with open('products.csv','r') as file:
    reader=csv.DictReader(file)
    for row in reader:
        for key,val in row.items():
            print(f'{key}:{val}')
        print('\n')
                         
```

```python
new_product=[{
        'Product Name':'mobile',
        'Product Brand':'samsung',
        'Product Price':'$699.99',
    }]
```

```python
file_name=input('Enter your file name')
file_exist=os.path.exists(file_name) and os.path.getsize(file_name)>0
with open(f'{file_name}.csv','a',newline='') as file:
    columns=['Product Name','Product Brand','Product Price']
    writer=csv.DictWriter(file,fieldnames=columns)
    if not file_exist:
        writer.writeheader()
    writer.writerows(new_product)
```
