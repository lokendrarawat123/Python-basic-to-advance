### Login  Authentication System


```python
'''
Login User - if both usernaje and password is correct .
Invalid Password - if password is incorrect.
Invalid Username - if username is incorrect 
Both username or password Invalid - if both username and password is incorrect.
'''
```

```python
username=abc@gmail.com
password='1234567'
uname=input('Enter your Username:')
pwd=input('Enter your Password:')
if uname==username and pwd==password:
    print('login succuess')
elif uname==username and pwd!=password:
    print('Incorrect Password')
elif uname!=username and pwd==password:
    print('User not found with given username')
else:
    print('Username and password not correct')
    
```

### Nested if

```python
username='abc@gmail.com'
password='1234567'
uname=input('Enter your username:')
pwd=input('Enter your Password:')
if uname!=username and pwd!=password:
    print("Username and Password is Incorrect")
else:
    if uname==username:
        if pwd==password:
            print('login succuess')
        else:
            print('Incorrect Password')
    else:
        print('User not found with given username')
    
```

### ATM Amount Withdrawl

```python
'''
Conditions
-----------
1.Pin must be correct
2.Balance must be less then availabe balance.
3.Withdraw balance must be multiple of 1000s.
'''
```

```python
pin=1234
available_balance=500009
upin=int(input('Enter your pin :'))

if upin==pin:
    withdrawl_balance=int(input('Enter withdrawl balance:'))
    if available_balance>withdrawl_balance:
        if withdrawl_balance%1000==0:
            print("Amount successfully withdraw.Thank You.")
        else:
            print('Amount should be multiple of 1000s.')
    else:
        print('Not enough balance')
else:
    print('Incorrect pin')
```
