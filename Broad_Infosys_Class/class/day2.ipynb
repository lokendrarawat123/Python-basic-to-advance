```python
anual_salary=int(input("enter anual salary:"))
tax_amount=0
if anual_salary<0:
    print("invalid amount")
else:
    if anual_salary>=0 and anual_salary<=250000:
        tax_amount=0
    elif anual_salary>250000 and anual_salary<=500000:
        tax_amount=(anual_salary*10)/100
    elif anual_salary>500000 and anual_salary<=1000000:
        tax_amount=(anual_salary*15)/100
    elif anual_salary>1000000:
        tax_amount=(anual_salary*30)/100

    print(f'your anual salary is = Rs.{anual_salary}')
    print(f'taxable income = Rs.{tax_amount}')
    print(f'remaining amount = Rs.{anual_salary-tax_amount}')
    

```
