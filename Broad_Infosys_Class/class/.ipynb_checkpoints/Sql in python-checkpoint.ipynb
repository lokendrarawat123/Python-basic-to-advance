```python
cursor
```

# Create Table

## Customer Table

```python
try:
    cursor.execute('''
    Create Table Customers(
    customer_id integer primary key autoincrement,
    customer_name text not null,
    contact text not null ,
    email text not null unique,
    customer_type text not null,
    constraint chk_customer_type Check(customer_type in ('Vip','Regular','Non-Regular'))
    );
    
    ''')
    print('Costumer table created succesfully')
except sqlite3.OperationalError as e:
    print(e)
```

## Staff Table

```python
sql='''
    create table Staffs(
        staff_id integer primary key autoincrement,
        staff_name text not null,
        staff_phone text not null unique,
        staff_email text not null unique,
        staff_position text  not null,
        staff_type text not null,
        constraint chk_staff_position check(staff_position in ('Senior','junior','Associate')),
        constraint chek_staff_type check(staff_type in ('Full-Time','Part-Time','Contract'))
        );
    
'''

try:
    cursor.execute(sql)
    print('Staff table created succefully')
except sqlite3.OperationalError as e:
    print(e)
```

# Order Table

```python
sql='''
Create Table Orders(
    order_id integer primary key autoincrement,
    order_date date not null,
    order_status not null,
    order_destination not null,
    customer_id integer,
    staff_id integer,
    constraint che_order_status check(order_status in ('Completed','Pending','Rejected','Processing')),
    foreign key(customer_id) references customers (customer_id),
    foreign key(staff_id) references staffs (staff_id)
    
    

);
'''
try:
    cursor.execute(sql)
    print('Order table  crated succesfully')
except sqlite3.OperationalError as e:
    print(e)
    
    
```

```python
'''

Insert ,Update ,Delete (x-Lock),Select(S-Lock)
--- Commit -> To Change data Permanently
--- Rollback -> Undo operation

--Database Concurrency(Deadlock)
select * from cusromers;
select * from staff;
we can't  do insert and update or delete in same time
'''
```

## Insert in staff tables


```python

# Sample data: (staff_name, staff_phone, staff_email, staff_position, staff_type)
staffs_data = [
    ("Aarav Adhikari", "9851011111", "aarav@company.com", "Senior", "Full-Time"),
    ("Binita Shrestha", "9841022222", "binita@company.com", "junior", "Part-Time"),
    ("Kiran KC", "9801033333", "kiran@company.com", "Associate", "Contract"),
]

try:
    cursor.executemany(
        """
        INSERT INTO Staffs (staff_name, staff_phone, staff_email, staff_position, staff_type)
        VALUES (?, ?, ?, ?, ?);
    """,
        staffs_data,
    )

    conn.commit()
    print("Staff records successfully inserted!")

# except sqlite3.IntegrityError as e:
#     print(
#         f"Data insert huna sakena (Phone/Email unique vaena ki ya CHECK constraint milena): {e}"
#     )
except sqlite3.OperationalError as e:
    print(f"Operational error: {e}")
```
