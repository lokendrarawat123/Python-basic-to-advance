# Inheritance

```python
'''
-Parent Class 
-Child Class
1.Child Class inherits/extennds Parent class.
2.if chils class inherits parent class then child class must inherits parent class constructor.-->super()
'''
```

```python
# parent Class
class Customer:
    def __init__(self,customer_id,customer_name,customer_address,customer_phone,customer_email):
        self.customer_id=customer_id
        self.customer_name=customer_name
        self.customer_address=customer_address
        self.customer_phone=customer_phone
        self.customer_email=customer_email
    def customerDetails(self):
        print(f'Customer Id :{self.customer_id}')
        print(f'Customer Name:{self.customer_name}')
        print(f'Customer Address:{self.customer_address}')
        print(f'Costumer Phone:{self.customer_phone}')
        print(f'Costumer Email:{self.customer_email}')
        
    
```

```python
# Child Class
class Order(Customer):
    def __init__(self,order_id,order_date,order_destination,order_status,customer_id,customer_name,customer_address,customer_phone,customer_email):  
        super().__init__(customer_id,customer_name,customer_address,customer_phone,customer_email)
        self.order_id=order_id
        self.order_date=order_date
        self.order_destination=order_destination
        self.order_status=order_status
    def orderDetails(self):
        print(f'Order Id:{self.order_id}')
        print(f'Order Date:{self.order_date}')
        print(f'Order Destination:{self.order_destination}')
        print(f'Order Status:{self.order_status}')
    
        
        
```

```python
orders=[]
```

```python
o = Order(
    order_id=5002,
    order_date="2026-08-23",
    order_destination="Lalitpur, Nepal",
    order_status="Processing",
    customer_id=102,
    customer_name="kamal Thapa",
    customer_address="Kathmandu",
    customer_phone="9801234567",
    customer_email="kamal@gmail.com")
orders.append(o)
```

```python
class Category:
    def __init__(self,category_id,category_name):
        self.category_id=category_id
        self.category_name=category_name
    def categoryDetails(self):
        print(f'Category Id :{self.category_id}')
        print(f'Category Name:{self.category_name}')
    
```

```python
## Multi Level  Inheritance
```

```python
# parent Class
class Customer:
    def __init__(self,customer_id,customer_name,customer_address,customer_phone,customer_email):
        self.customer_id=customer_id
        self.customer_name=customer_name
        self.customer_address=customer_address
        self.customer_phone=customer_phone
        self.customer_email=customer_email
    def customerDetails(self):
        print(f'Customer Id :{self.customer_id}')
        print(f'Customer Name:{self.customer_name}')
        print(f'Customer Address:{self.customer_address}')
        print(f'Costumer Phone:{self.customer_phone}')
        print(f'Costumer Email:{self.customer_email}')
        
    
```

```python
# Child Class
class Order(Customer):
    def __init__(self,order_id,order_date,order_destination,order_status,customer_id,customer_name,customer_address,customer_phone,customer_email):  
        super().__init__(customer_id,customer_name,customer_address,customer_phone,customer_email)
        self.order_id=order_id
        self.order_date=order_date
        self.order_destination=order_destination
        self.order_status=order_status
    def orderDetails(self):
        print(f'Order Id:{self.order_id}')
        print(f'Order Date:{self.order_date}')
        print(f'Order Destination:{self.order_destination}')
        print(f'Order Status:{self.order_status}')
    
        
        
```

```python
class Order_Items(Order):
    def __init__(self,item_id,item_quantity,item_price,item_desc,order_id,order_date,order_destination,order_status,customer_id,customer_name,customer_address,customer_phone,customer_email):
        super().__init__(order_id,order_date,order_destination,order_status,customer_id,customer_name,customer_address,customer_phone,customer_email):
        self.item_id=item_id
        self.item_quantity=item_quantity
        self.item_price=item_price
        self.item_desc=item_desc
    def orderItem_Details(self):
        print(f'Item Id:{self.item_id}')
        print(f'Item Quantity:{item_quantity}')
        print(f'Item Price:{item_price}')
        print(f'Item Description:{item_desc}')
        
              
```

## Multipler Inheritance:


```python
class Category:
    def __init__(self,category_id,category_name):
        self.category_id=category_id
        self.category_name=category_name
    def category_details(self):
        print(f'Category Id :{self.category_id}')
        print(f'Category Name :{self.category_name}')
        
```

```python
class Brand:
    def __init__(self,brand_id,brand_name):
        self.brand_id=brand_id
        self.brand_name=brand_name

    def brand_details(self):
        print(f'Brand ID :{self.brand_id}')
        print(f'Brand Name :{self.brand_name}')
```

```python
class Product(Category,Brand):
    def __init__(self,product_id,product_name,product_desc,product_price,category_id,category_name,brand_id,brand_name):
        ## accessing both category and brand class constructor 
        Category.__init__(self,category_id,category_name)
        Brand.__init__(self,brand_id,brand_name)
        self.product_id=product_id
        self.product_name=product_name
        self.product_desc=product_desc
        self.product_price=product_price
    def product_details(self):
        print(f'Product ID:{self.product_id}')
        print(f'Product Name:{self.product_name}')
        print(f'Product Description:{self.product_desc}')
        print(f'Product Price:{self.product_price}')
        
        
```

```python
obj=Product('1','Car','this is product description','Rs.34444444','101','Tesla','201','vehicle')
obj.category_details()
```

```python
obj.brand_details()
```

```python
obj.product_details()
```

# Polymorphism

```python
# Ducktyping
```

```python
class Messenger:
    def message(self,send_to ,send_from,message):
        print('Messenger')
        print(f'Sender:{send_from}')
        print(f'Reciever:{send_to}')
        print(f'Message:{message}')
class Viber:
    def message(self,send_to ,send_from,message):
        print('Viber')
        print(f'Sender:{send_from}')
        print(f'Reciever:{send_to}')
        print(f'Message:{message}') 
       
class Whatsapp:
    def message(self,send_to ,send_from,message):
        print('Whatsapp')
        print(f'Sender:{send_from}')
        print(f'Reciever:{send_to}')
        print(f'Message:{message}')
        
class Slack:
    def message(self,send_to ,send_from,message):
        print('Slack')
        print(f'Sender:{send_from}')
        print(f'Reciever:{send_to}')
        print(f'Message:{message}')
        
```

```python
def send_message(obj,send_to,send_from,message):
    obj.message(send_to,send_from,message)

```

```python
send_message(Messenger(),'shyam','ram','Hello Ram')
```

```python
send_message(Whatsapp(),'shyam','ram','Hello Ram')
```

# Abstraction

```python
'''
1.Parent class must be abstract class.
2.Atleast one function in abstract calss must be abstract function.
3.Child class implements Parent class data.
'''
```

```python
# ABC --> Abstract Base Class
from abc import ABC,abstractmethod
```

```python
class Bank(ABC):
    @abstractmethod
    def user(self):
        return "******"
    def connection(self):
        return 'Connected to server...'
```

```python
class MobileBanking(Bank):
    def statement(self):
        return 'User Statement'
    def user(self):
        return f'username:Arthur'
```

```python
mb=MobileBanking()
```

```python
mb.user()
```

```python
# API Key = “ABC123456789XYZ
class Credential:
    def __init__(self,api_key):
        self.__api_key=api_key
    def get_masked_key(self):
        masked_key=self.__api_key.replace(self.__api_key[3:-3],'************')
        print(masked_key)
        
        


        
```

```python
# API Key = “ABC123456789XYZ
class Credential:
    def __init__(self,api_key):
        self.__api_key=api_key
    def get_masked_key(self):
        first_3=self.__api_key[:3]
        last_3=self.__api_key[-3:]
        total_char=len(self.__api_key)-6
        asterisk='*'*total_char
        masked_value=f'{first_3}{asterisk}{last_3}'
        print(masked_value)
        
        


        
```
