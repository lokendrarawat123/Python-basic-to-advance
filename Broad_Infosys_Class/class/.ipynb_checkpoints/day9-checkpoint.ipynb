All arguments (*args):
What is *args?
Imagine you are making a juice blend. Sometimes you want to use 2 fruits, and other times you want to use 5 fruits.

Normally, if a Python function expects 2 items, you must give it exactly 2 items. If you give it 3 or 1, Python gives you an error.

*args solves this problem. It tells Python: "I don't know how many items the user will give me, so accept them all and put them in a list-like collection (a tuple)."

Understanding the Syntax
The Star (*): This is the magic part. The * tells Python to gather all extra inputs into a single group.

The Word args: Short for "arguments". You can technically name it anything (like *fruits), but programmers usually stick to *args so everyone understands the code.


```python
def empDetails_args(*empdetails):
    for val in empdetails:
        print(val)
```

```python
empDetails_args(1,'Bob')
```

```python
## if we need to add compulsary argument then we can do this following.

def empDetails_args(id,*empdetails):
    print(id)
    for val in empdetails:
        print(val)
empDetails_args(1)
```

```python
def make_shopping_list(*items):
    print("Your Shopping List:")
    for item in items:
        print(f"- {item}")

# Call it with 2 items
make_shopping_list("Apples", "Milk")

print("---")

# Call it with 4 items
make_shopping_list("Bread", "Eggs", "Cheese", "Butter")
```

```python
 ### All keyword Argument
```

```python
def empDetails_kwargs(**kwargs):
    print(kwargs)
    print()
    for key,val in kwargs.items():
        print(f'{key}:{val}')
```

```python
empDetails_kwargs(id=1,name='Bob',position='Senior')
```

```python
def add(a,b):
    return f'addition of {a} and {b} is :{a+b}'
add(3,5)
```

```python
def odd_even(val):
    if val%2==0:
        print('even')
    else:
        print('odd')
odd_even(5)
```

```python
## in lambda function
odd_ev=lambda val: f'{val}:even' if val%2==0 else f'{val}:odd'
print(odd_ev(7))
print(odd_ev(4))
```

```python
def odd_even():
    for i in range(1,21):
        if i%2==0:
            print('even')
        else:
            print('odd')

odd_even()
```

```python
for_od_ev=lambda:[f'{i}:Even' if i%2==0 else f'{i}:odd' for i in range (1,21)]
for_od_ev()
```

### Nested Function

```python
def add(a,b):
    return f'addition of {a} and {b} is = {a+b}'
def sub(a,b):
    return f'subtraction of {a} and {b} is = {a-b}'
def multi(a,b):
    return f'multiplication of {a} and {b} is = {a*b}'
def div(a,b):
    return f'division of {a} and {b} is = {a/b}'
```

```python
def calculator():
    x=int(input('Enter first number:'))
    y=int(input('Enter second number: '))
    choice=int(input('Enter your choice:'))
    if choice==1:
        print(add(x,y))
    elif choice==2:
        print(sub(x,y))
    elif choice==3:
        print(multi(x,y))
    elif choice==4:
        print(div(x,y))
    else:
        print('Invalid Choice')
```

```python
calculator()
```
