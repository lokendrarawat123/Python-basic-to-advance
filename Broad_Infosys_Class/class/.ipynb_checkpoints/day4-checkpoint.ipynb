```python
#  write a program to print first 10 natural number:
for i in range(1,11,1):  #(1,10,2)
    print(i)
```

```python
## write a progaram to check odd and even numbers for 20 natural numbers.
for i in range(21):
    if i%2==0:
        print(f'{i}-even')
    else:
        print(f'{i}-odd')
```

```python
## User login system 
# user can  attempt 3 only TIMES
username="abc@gmail.com"
password='1234567'
attempts=3
for i in range(3):
    print(f'Remaining attempts:{attempts}')
    uname=input('Enter your Username:')
    pwd=input('Enter your Password:')
    if uname!=username and pwd!=password:
        print("Username and Password is Incorrect")
        attempts-=1
    else:
        if uname==username:
            if pwd==password:
                print('login succuess')
                break
            else:
                print('Incorrect Password')
                attempts-=1
        else:
            print('User not found with given username')
            attempts-=1
    
else:
    print('Login Attempt Limit.Account Blocked.')
    
```

### Break and Continue.


```python
'''
-Break--> To Stop program forcefully.
Continue--> To skip value /continue
'''
```

```python
for i in range(1,11):
    if(i==5):
       break
    print(i)
```

```python
for i in range(1,11):
    if(i==5):
       continue
    print(i)
```

```python
## Write a progra that loops though numbers from 1 u to 30 .
# if a number is a multiple of 3,skip printing entirely usring
#the continue statement.print all other numbers

for i in range(1,31):
    if i%3==0:
        continue
    print(i)
```

```python
text='Hello'
has_L=False
for word in text:
    if word=='l':
        has_L=True
    else:
        has_L=False
    print(has_L)
```
