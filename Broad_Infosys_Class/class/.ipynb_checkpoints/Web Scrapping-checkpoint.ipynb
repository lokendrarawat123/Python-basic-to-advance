```python
from selenium import webdriver #-->access web browser
from selenium.webdriver.common.by import By # access web site elements/tags
```

```python
website="https://merolagani.com/LatestMarket.aspx"
```

```python
# access browser
# driver=webdriver.Edge()
driver=webdriver.Chrome()
```

```python
driver.get(website)
```

## Access Table Body


```python
# find element(By.finders,element_name)

table_body=driver.find_element(By.TAG_NAME,'tbody')
# table_body
```

```python
table_rows=table_body.find_elements(By.TAG_NAME,'tr')
# table_rows
```

```python
# for row in table_rows:
#     columns=row.find_elements(By.TAG_NAME,'td')
#     for col in columns:
#         print(col.text)
    
```

```python
symbol_col=[]
ltp_col=[]
change_col=[]
open_col=[]
high_col=[]
low_col=[]
qty_col=[]
pclose_col=[]
diff_col=[]
```

```python
# append each value in specefic column
for row in table_rows:
    columns=row.find_elements(By.TAG_NAME,'td')
    symbol_col.append(columns[0].text)
    ltp_col.append(columns[1].text)
    change_col.append(columns[2].text)
    open_col.append(columns[3].text)
    high_col.append(columns[4].text)
    low_col.append(columns[5].text)
    qty_col.append(columns[6].text)
    pclose_col.append(columns[7].text)
    diff_col.append(columns[8].text)
```

```python
# print(f'{pclose_col}')
```

```python
data={
    'Symbol':symbol_col,
    'ltp':ltp_col,
    '%Change':change_col,
    'Open':open_col,
    'High':high_col,
    'Low':low_col,
    'Qty':qty_col,
    'PClose':pclose_col,
    'Diff':diff_col
}
```

```python
driver.quit()
```

```python
import pandas as pd
```

```python
df=pd.DataFrame(data)
# df
```

```python
'''
to_csv()
to_excel()
to_json()
'''
```

```python
df.to_csv('share.csv',index=False)
df.to_excel('share.xlsx',index=False)
df.to_json('share.json',indent=4)
```
