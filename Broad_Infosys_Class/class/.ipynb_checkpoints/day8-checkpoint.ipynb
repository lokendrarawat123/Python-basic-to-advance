```python
'''Lab Question 3: Student Database Update Create a dictionary named student with the 
following keys: "name": "Alice", "age": 20, and "major": "Physics".
1. Add a new key-value pair: "gpa": 3.8.
2. Update the "major" to "Computer Science".
3. Delete the "age" key using the pop() method.
4. Loop through the dictionary and print only the keys.
5. Loop through the dictionary using .items() and print the details in the format: Field: 
Value.
'''
```

```python
student={
    'name':'Alice',
    'age':20,
    'major':'Physics'

}
# 1
student['gpa']=3.8
student
```

```python
# 2 
student['major']='Computer Science'
student
```

```python
# 3
student.pop('age')
student
```

```python
# 4
for key in student.keys():
    print(key)
```

```python
# 5
for key , value in student.items():
    print(f'{key}:{value}')
```

```python
# Lab Question 7: The Gradebook (Dictionary with Nested Lists) Create a dictionary 
# named gradebook: gradebook = {"Sam": [80, 85, 90], "Tara": [92, 88, 95]}
# 1. Add a new student named "Leo" with scores [70, 75, 80].
# 2. Update Sam's second score (index 1) from 85 to 88.
# 3. Delete the last score in Tara's list using the pop() method.
# 4. Use a for loop and .items() to print each student's name and their highest score 
# using the max() function
```

```python
gradebook={
"Sam": [80, 85, 90],
"Tara": [92, 88, 95]   
}
```

```python
gradebook["Leo"]=[70, 75, 80]
gradebook
```

```python
gradebook['Sam'][1]=88
gradebook
```

```python
gradebook['Tara'].pop()
gradebook
```

```python
for key,val in gradebook.items():
    print(f'{key}:{max(val)}')
```
