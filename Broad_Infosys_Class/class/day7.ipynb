```python
prices =[100,250,75,400,120,310]
cheap_items=[]
expensive_items=[]
# for price in prices:
#     if price>200:
#         expensive_items.append(price)
#     else:
#         cheap_items.append(price)
for price in prices :
    expensive_items.append(price) if price>=200 else cheap_items.append(price)
print(cheap_items)
print(expensive_items)
print('total cheap items are =',len(cheap_items))
print('total expensive items are =',len(expensive_items))
```

```python
data =[1,2,2,2,3,3,1,4,5,5,5,6,8,22]
duplicate_data=[]
unique_data=[]
for i in data:
    if not i in unique_data:
        unique_data.append(i)
    else:
        duplicate_data.append(i)
print(duplicate_data)
print(unique_data)
        
```
