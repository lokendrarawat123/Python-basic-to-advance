```python
class PinError(Exception):
    pass
class Insufficient(Exception):
    pass
class IncorrectAmount(Exception):
    pass
```

```python
def atm():
    pin =1234
    balance=50000
    user_pin=int(input('Enter your pin'))
    if user_pin==pin:
        amount=int(input('Enter your withdrawl amount'))
        if amount<balance:
            if amount%1000==0:
                print(f'Rs.{amount} has been withdrawn from A/C no.543263434543')
            else:
                raise IncorrectAmount('Amount Error:Amount must be multiple of 1000')
        else:
            raise Insufficient('Balance Error:withdraw amount must be less than available balance')
    else:
         raise PinError('Pin Error: Incorrect pin')
    
```

```python
try:
    atm()
except PinError as pe:
    print(pe)
except Insufficient as ie:
    print(ie)
except IncorrectAmount as ine:
    print(ine)
except ValueError as e:
    print('Amount must be integer')
    
```

```python
import random
```

```python
class DepositAmountError(Exception):
    pass
class WithdrawAmountError(Exception):
    pass
class AccountNotFoundError(Exception):
    pass
```

```python
## storage
accounts=[]
```

```python
class Bank:
    def __init__(self,name,initial_balance,job_position):
        self.name=name
        self.initial_balance=initial_balance
        self.job_position=job_position
        self.account_number=self.name[1:3]+''.join(str(random.randint(0,9)) for i in range(16))+self.job_position[1:3]
    def depositeAmount(self,amount):
        if amount>100:
            self.initial_balance+=amount
        else:
            raise DepositeAmountError('Amount Error: Deposit amount must be more than Rs.100')
    def withdrwaAmount(self,amount):
        if amount <self.initial_balance:
            self.initial_balance-=amount
            print(f'Rs.{amount} has been withdrwan from A/C no.{self.account_number}')
        else:
            raise WithdrawAmountError('Amount Error: Withdraw amount must be less than available balance.')
    def userDetails(self):
        print('User Details')
        print('='*30)
        print(f'Account Holder Name:{self.name}')
        print(f'Account Number : {self.account_number}')
        print(f'Balance Amount : Rs.{self.initial_balance}')
        print(f'Job Position : {self.job_position}')
```

```python
b=Bank('Shyam',4000,'Developer')
accounts.append(b)
b.userDetails()
```

```python
def findAccountbyAccountNumber(acc_number):
    for account in accounts:
        if account.account_number==acc_number:
            return account
    else:
        raise AccountNotFoundError('Account Number Error: Account not found with given account number does not found.')
```
