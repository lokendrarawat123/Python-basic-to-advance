```python
try:
    a=int(input('Enter a number'))
    b=int(input('Enter a number'))
    c=a/b
    print(c)
except ValueError as e:
    print(f'Error: Only number values are accepted by input field')
except ZeroDivisionError as e:
    print(f'Error: Number can not divsion by 0 ')
    
```
