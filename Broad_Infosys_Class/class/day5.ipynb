```python
# Automated password  compiler validator
'''
1.Password must be min 8 characters long and max 16  character long.
-> special Characters -> Regular Expressions(Regex)
2.Password must contain atleast 1 uppercase,1 lowercase ,1 special charcters,1 digit/number.
3.Password must not contain any space between.
'''
```

```python
password=input('Enter your password')
pwd_length=len(password)
is_upper=False
is_lower=False
is_digit=False
is_special=False
is_space=False
special = "!\"#$%&'()*+,-./:;<=>?@[\\]^_`{|}~"
if pwd_length>=8 and pwd_length<=16:
    for word in password:
        if word.isupper(): # check upper case
            is_upper=True
        if word.islower():
            is_lower=True
        if word.isdigit():
            is_digit=True
        if word==" ":
            is_space=True
        if word in special :
            is_special=True
        
            
            
else:
    print('Password must be minimum of 8 characters long or maximum of 16 charaters long.')
if is_space:
    print('password must not contain any space character')
if not is_upper:
    print('Password must contain atleast 1 uppercase character.')
if not is_lower:
    print('Password must contain atleast 1 lowercase character.')
if not is_digit:
    print('Password must contain atleast 1 digit character.')
if not is_special:
    print('Password must contain atleast 1 special character.')
    
```
