```python
# using while loop
number=int(input("enter a number"))
sum=0
quotient=number
while quotient>0:
    remainder=number%10
    quotient =number//10
    sum+=remainder
    number=quotient
print(f"sum= {sum}")
    
    
```

```python
number = int(input("Enter a number: "))
sum = 0
quotient = number

while quotient > 0:
    remainder = quotient % 10  # Use quotient instead of number
    quotient = quotient // 10   # Use quotient instead of number
    sum += remainder

print(f"Sum of digits = {sum}")
```
