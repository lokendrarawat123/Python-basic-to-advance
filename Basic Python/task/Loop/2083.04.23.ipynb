```python
for i in range(5):          # Row control (0 dekhi 4)
    for j in range(5):      # Column control (0 dekhi 4)
        if j < i:
            print(" ", end="")  # Space print garne
        else:
            print("x", end="")  # 'x' print garne
    print()                 # Line change garne
```

```python
# write a program to check a number prime or not.
number = int(input("Enter a number: "))

if number <= 1:
    print("Neither prime nor composite")
else:
    for i in range(2, number):
        if number % i == 0:
            print("Composite")
            break
    else:
        print("Prime")
```

```python
rows = 4
for i in range(1, rows + 1):
    print("  " * (rows - i) + "* " * (2 * i - 1))
```

```python
n = 4
# Top half
for i in range(n, 0, -1):
    print("* " * i)
# Bottom half
for i in range(2, n + 1):
    print("* " * i)
```

```python
rows = 5
num = 1
for i in range(1, rows + 1):
    for _ in range(i):
        print(num, end=" ")
        num += 1
    print()
```

```python
rows = 5
for i in range(1, rows + 1):
    for j in range(1, i + 1):
        print((i + j + 1) % 2, end="")
    print()
```

```python
rows = 5
for i in range(1, rows + 1):
    for j in range(1, i + 1):
        if j == 1 or j == i or i == rows:
            print("* ", end="")
        else:
            print("  ", end="")
    print()
```

```python
rows, cols = 5, 6
for i in range(1, rows + 1):
    for j in range(1, cols + 1):
        if i == 1 or i == rows or j == 1 or j == cols:
            print("*", end="")
        else:
            print(" ", end="")
    print()
```
