### WAP to search for a name in a list and display the total number of occurrences (case-insensitive).

```python
names = ["Alice", "Bob", "charlie", "ALICE", "David", "alice"]

search_name = input("Enter a name to search: ")

count = 0

# Loop through each name in the list
for name in names:
    # Convert both names to lowercase to make it case-insensitive
    if name.lower() == search_name.lower():
        count += 1

# Display result
if count > 0:
    print(f"'{search_name}' was found {count} time(s) in the list.")
else:
    print(f"'{search_name}' was not found in the list.")
```

## WAp to create a list of numbers and find the second largest number.

```python
numbers = [12, 45, 2, 99, 45, 87, 99, 23]

# Step 1: Remove duplicates using set() so top numbers aren't identical
unique_numbers = list(set(numbers))

# Step 2: Sort the numbers in ascending order
unique_numbers.sort()

# Step 3: Access the second element from the end (-2 index)
if len(unique_numbers) >= 2:
    second_largest = unique_numbers[-2]
    print(f"The second largest number is: {second_largest}")
else:
    print("List does not have at least two unique numbers.")
```

```python
numbers = [12, 45, 200, 99, 87]

largest = 0
second_largest = 0

for num in numbers:
    if num > largest:
        second_largest = largest
        largest = num
    elif num > second_largest and num != largest:
        second_largest = num

print(second_largest)  # Output: 87 (Sahi kaam garchha!)
```

WAP to create a list of numbers and separate even and odd numbers into two different lists.

```python
# Initial list of numbers
numbers = [12, 7, 19, 24, 33, 40, 55, 68, 81, 90]

# Empty lists to store separated numbers
even_numbers = []
odd_numbers = []

# Loop through each number in the list
for num in numbers:
    if num % 2 == 0:
        even_numbers.append(num)  # Add to even list
    else:
        odd_numbers.append(num)   # Add to odd list

# Display the results
print("Original List:", numbers)
print("Even Numbers: ", even_numbers)
print("Odd Numbers:  ", odd_numbers)
```

WAP to create a list of numbers and remove duplicate elements.

```python
numbers = [10, 20, 30, 20, 40, 10, 50, 30, 60]

unique_numbers = []

for num in numbers:
    if num not in unique_numbers:
        unique_numbers.append(num)

print("Original List:", numbers)
print("Without Duplicates:", unique_numbers)
```
