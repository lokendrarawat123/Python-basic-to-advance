### 1. WAP to check whether the given password is strong or weak.Password must contain at least 10 chars, 2 uppercase, 1 number, 1 special character.


```python
password=input("enter your password")
number=False
count=0
special_character=False
for i in password:
        if i.isupper():
           count+=1
        if not (i.isalnum()):
            special_character=True
        if(i.isdigit()):
            number=True
if(len(password)>=10 and count>=2 and number and special_character):
    print("strong password")
else:
    print("weak password")
print(count,number,special_character)
```

## 2.WAP to check whether an email is valid.
###  Condition:
    Contains exactly one '@'
    Contains at least one '.'

```python
email=input("enter your gmail")

at_count = 0
dot_count = 0

for char in email:
    if char == '@':
        at_count += 1
        
    elif char == '.':
        dot_count += 1
        

if at_count == 1 and dot_count >= 1:
    print(f"'{email}'email  is Valid")
else:
    print(f"'{email}' email is Invalid")
    
```

## WAP to check whether a URL is valid.
## Condition:
    Contains at least one .
    Starts with www




```python
url = input("Enter a URL: ")


starts_with_www = url.startswith("www")


has_dot = "." in url

if starts_with_www and has_dot:
    print("Valid URL")
else:
    print("Invalid URL")
```

## WAP to check whether a mobile number is valid.
## Condition:
    Exactly 10 digits


```python
mobile = input("Enter your mobile number: ")

# Condition 1: Check if all characters are digits
# Condition 2: Check if the length is exactly 10
if mobile.isdigit() and len(mobile) == 10:
    print("Valid Mobile Number")
else:
    print("Invalid Mobile Number")
```
