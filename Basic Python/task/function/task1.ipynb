```python
# 3 - WAP using a function to analyze bank account transactions. 
# The function should calculate total deposit, total withdrawal, final balance, and account status.
# Initial balance = Rs. 10,000
# If final balance ≥ 5,000 → "Active"
# Otherwise → "Low Balance"
```

```python
def bank_account(transactions):
    initial_balance = 10000
    total_deposit = 0
    total_withdrawal = 0

    for amount in transactions:
        if amount > 0:
            total_deposit = total_deposit + amount
        else:
            total_withdrawal = total_withdrawal + (amount * -1)

    final_balance = initial_balance + total_deposit - total_withdrawal

    if final_balance >= 5000:
        status = "Active"
    else:
        status = "Low Balance"

    print(f"Total Transactions Count : {len(transactions)}")
    print(f"Total Deposit            : Rs. {total_deposit}")
    print(f"Total Withdrawal         : Rs. {total_withdrawal}")
    print(f"Final Balance            : Rs. {final_balance}")
    print(f"Account Status           : {status}")



transactions = [5000,-2000,1500,-3000,8000,-4500,-1000,2500,-6000,-2000   
]


bank_account(transactions)
```

```python
# 5- WAP to calculate an employee's net salary using nested functions.
# bonus() → 10% if salary ≥ 50000, else 5%.
# tax() → 8% if gross salary > 60000, else 3%.
# Calculate and display Gross Salary and Net Salary.

def net_salary(basic_salary):
    def bonus(salary):
        if salary >= 50000:
            return salary * 0.10  
        else:
            return salary * 0.05 
    bonus_amount = bonus(basic_salary)
    gross_salary = basic_salary + bonus_amount

    def tax(gross):
        if gross > 60000:
            return gross * 0.08  
        else:
            return gross * 0.03  

    tax_amount = tax(gross_salary)

    net_salary = gross_salary - tax_amount

    print(f"Basic Salary  : Rs. {basic_salary:.2f}")
    print(f"Bonus Amount  : Rs. {bonus_amount:.2f}")
    print(f"Gross Salary  : Rs. {gross_salary:.2f}")
    print(f"Tax Amount    : Rs. {tax_amount:.2f}")
    print(f"Net Salary    : Rs. {net_salary:.2f}")


emp_salary = float(input("Enter employee basic salary: "))

net_salary(emp_salary)
    
```
