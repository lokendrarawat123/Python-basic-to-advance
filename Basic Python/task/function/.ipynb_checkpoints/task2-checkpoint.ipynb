```python
# write a python function to multiply all the numbers in list.
my_list=[1,2,3,4,5,6,7]
def cal_multiply(list1):
    total=1
    for number in list1:
        total*=number
    return total
multi_sum=cal_multiply(my_list)
print(multi_sum)
        
```

```python
# write a python function to find factorial of a given non negative number.
# 5*4*3*2*1
n=int(input('Enter a number to find factorial number: '))
def find_factorial(n):
    total=1
    if n==0 or n==1:
        return total
    else:
        for i in range(n,1,-1):
            total*=i
    return total

if n<0:
    print('Enter positive number: ')
else:
    sum=find_factorial(n)
print(sum)
    
```

```python
# write a python code to reverse a string:
text="abcdefghi"
def text_reverse(text):
    reversed_text = text[::-1]
    return reversed_text

reverse_txt=text_reverse(text)
print(reverse_txt)
```

```python
def reverse_string(s):
    reversed_s = ""
    for char in s:
        reversed_s = char + reversed_s
    return reversed_s

text = input("Enter a string: ")
print("Reversed string:", reverse_string(text))
```

```python
# write a python function that accept string and calculate the number of uppercase letter and  lowercase letter.
def count_st(text):
    uppercase_letter=[]
    lowercase_letter=[]
    for i in text:
        if i.isupper():
            uppercase_letter.append(i)
        elif i.islower():
            lowercase_letter.append(i)
        else:
            pass
    print(f'uppercase letter list={uppercase_letter}')
    print(f'lower case letter list ={lowercase_letter}')
text=input('Enter a text')
count_st(text)
            
```

```python
# Write a python function that takes a list and returns a new list with unique elements of the first list.
def get_unique_elements(lst):
    unique_lst = []
    for item in lst:
        if item not in unique_lst:
            unique_lst.append(item)
    return unique_lst

numbers = [1, 2, 2, 3, 4, 4, 5,6,7,5,8,5,8,4,0,5,6,4,]
print("Unique list:", get_unique_elements(numbers))
```

```python
# Write a Python function that takes a number as a parameter and check the number is prime or not.
def is_prime(n):
    for i in range(2, n):
        if n % i == 0:
            return False
    return True


num = int(input("Enter a number: "))

if num < 2:
    print("Enter valid number (2 or greater).")
else:
    if is_prime(num):
        print(f"{num} is a Prime number.")
    else:
        print(f"{num} is Not a Prime number.")
```

```python
#  Write a python function to print the even numbers from a given list.
def print_even_numbers(lst):
    even_numbers = [num for num in lst if num % 2 == 0]
    print("Even numbers:", even_numbers)

my_list = [10, 15, 20, 25, 30]
print_even_numbers(my_list)
```
