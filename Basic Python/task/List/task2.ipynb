```python
# 1.WAP to store multiple users' name, address, age, and one or more phone numbers
# in a dictionary and display all the stored data at the end.
```

```python
# Multiple users को डेटा राख्ने Dictionary
users = {
    "user1": {
        "name": "Suman Shrestha",
        "address": "Kathmandu",
        "age": 22,
        "phone": ["9841000000", "9801000000"]  # Multiple phone numbers in a list
    },
    "user2": {
        "name": "Aayush Sharma",
        "address": "Pokhara",
        "age": 24,
        "phone": ["9851000000"]
    },
    "user3": {
        "name": "Pooja Karki",
        "address": "Lalitpur",
        "age": 21,
        "phone": ["9818000000", "01-5500000"]
    }
}

# सबैसँग Dynamic Input लिने र Display गर्ने कोड:
n = int(input("Enter number of users to add: "))
user_data = {}

for i in range(n):
    print(f"\n--- Enter details for User {i+1} ---")
    user_id = f"user_{i+1}"
    name = input("Enter Name: ")
    address = input("Enter Address: ")
    age = int(input("Enter Age: "))
    
    # Phone numbers हरूलाई comma (,) ले छुटाएर input लिने
    phones_input = input("Enter Phone Number(s) separated by comma: ")
    phones = [p.strip() for p in phones_input.split(",")]

    # Dictionary मा store गर्ने
    user_data[user_id] = {
        "name": name,
        "address": address,
        "age": age,
        "phone": phones
    }

# Display All Data
print("\n================ STORED USERS DETAILS ================")
for uid, info in user_data.items():
    print(f"\nID: {uid}")
    print(f"Name: {info['name']}")
    print(f"Address: {info['address']}")
    print(f"Age: {info['age']}")
    print(f"Phone(s): {', '.join(info['phone'])}")
```
