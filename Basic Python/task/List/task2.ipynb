```python
# wap to rotate a list the right be n positions and display the resulting list.
list1=[1,2,3,4,6,3,25,34]
list2=[]
n=int(input("enter  a number: "))
for i in range(n,len(list1)):
        list2.append(list1[i])
        if i==len(list1)-1:
            for j in range(0,n):
             list2.append(list1[j])
print(list2)
```
