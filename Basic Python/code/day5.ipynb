```python
# wap to calculate the electricity bill based on the number of units consumed .
# condition 
# if units consumed are 100 or less ,charge rs.5 per unit
# if units consumed are more than 100 up to 200 ,charge rs.8 per unit 
# if units consumed are more than 200 ,charge rs.10 per unit 

unit=int(input("unit:"))
bill= (unit<=100 )* unit*5 + (unit >=100 and unit<=200) * unit*8  + (unit>200)*unit*10
print(bill)
```
