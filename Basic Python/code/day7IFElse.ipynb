```python
# wap to progarm to create a simple calculator using operatior.

a=int(input("enter a number you want"))
operator=input("enter operator")
b=int(input("enter a number  you want"))


if(operator=="+"):
    print(f"the result is {a} + {b} = ",a+b) 
elif(operator=="-"):
    print(f"the result is {a} - {b} = ",a-b) 
elif(operator=="*"):
    print(f"the result is {a} * {b} = ",a*b) 
elif(operator=="/"):
    if(b==0):
        print("0 is not valid interger for divide")
    else:
        print(f"the result is {a} / {b} =" ,a/b) 
elif(operator=="**"):
    print(f"the result is {a} ** {b} =" ,a**b) 
elif(operator=="//"):
    print(f"the result is {a} // {b} =" ,a//b) 
elif(operator=="%"):
    if(b==0):
        print("0 is not valid for the modulas operation")
    print(f"the result is {a} % {b} =" ,a%b)
else:
    print("error")
    


          
```

```python

# wap to input marks of 5 subject,calculate total marks ,percentage,and grade.
# if any subject mark is below 32 ,display "failed"; otherwise display the result with grade.

subject1=float(input("enter the marks of subject1"))
subject2=float(input("enter the marks of subject2"))
subject3=float(input("enter the marks of subject3"))
subject4=float(input("enter the marks of subject4"))
subject5=float(input("enter the marks of subject5"))


if(subject1<32 or subject2<32 or subject3<32 or subject4<32 or subject5<32):
    print("failed")
else:
    total=subject1+subject2+subject3+subject4+subject5
    print(f"the total marks is {total} ")
    result=((subject1+subject2+subject3+subject4+subject5)/500)*100
    print(f"the percentage  is {result}% ")
    grade="E" 
    if(result>90):
        grade="A+"
    elif(result>80):
        grade="A"
    elif(result>70):
        grade="B+"
    elif(result>60):
        grade="B"
    elif(result>50):
        grade="C+"
    elif(result>40):
        grade="C"
    elif(result>30):
        grade="D+"
    elif(result>20):
        grade="D"
    else:
        grade="E"
    print(f"the result is pass with {result}%  with {grade} grade")

```

```python
# 1- WAP to check whether a given number is a prime number or not.
number=int(input("enter a number to check prime or not"))
if(number==0 or number==1):
    print(f"{number} is not prime nor even")
elif(number%2!=0 and number%3!=0 and number%5!=0):
    print("the number is prime")
else:
    print("the number is even")
```

```python
'''2- WAP to book movie tickets. Assume there are 100 tickets available.
If the requested number of tickets is available, calculate the total amount based
on the ticket class (VIP = Rs.1500, Normal = Rs.800). Apply a 10% discount
if the total amount is Rs.10,000 or more and display the final amount.
Otherwise, display "Tickets Not Available'''

ticket=int(input("enter the number of ticket"))
ticket_class=input("enter the which class do you want : \n 1 for vip and \n 2 for normal")
total_amount=0
if(ticket<=100):
    if(ticket_class=="1"):
        total_amount=ticket*1500
    elif(ticket_class=="2"): 
        total_amount=ticket*800   
    else:
        print(f"{ticket_class} ticket class is not available")
    if(total_amount>=10000):
        total_amount=total_amount-(total_amount*10)/100
        print(f"the total amount with discount = {total_amount:g}")
    else:
        print(f"the total amount is = {total_amount} ") 
else:
    print("ticket not available")
    
         
```

```python
'''
3- WAP to input consumed units and calculate electricity bill according to rules:
Up to 20 units ---- Rs. 80
21 - 30 units ------ Rs. 7 per unit
31-50 units ----- Rs. 8.5 per unit
Above 50 unit ----- Rs. 10 per unit.
'''
unit=int(input("enter total units"))
total_bill=0
if(unit>0 and unit<=20):
    total_bill=80
elif(unit>20 and unit<=30):
    total_bill=7*unit
elif(unit>30 and unit<=50):
    total_bill=8.5*unit
elif(unit>50):
    total_bill=10*unit
else:
    print("invalid unit")
print(f"total bill = {total_bill:.2f}")
    
```
