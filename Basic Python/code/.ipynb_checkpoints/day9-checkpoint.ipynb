```python
# while loop
 
#  while<condition>:
#     <operation>
# synatax 
i=0
while i<5:
     print(i)
     i+=1
```

```python
n=int(input("enter"))
i=1
while i<=10:
    print(f"{n}*{i}={n*i}")
    i+=1
```
