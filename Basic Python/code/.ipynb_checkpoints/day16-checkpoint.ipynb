```python
# function without arguments and no return type 
# function with arguments and no return type 
# function without arguments and  return type 
# function with arguments and  return type 
```

```python
# function without arguments and no return type \
def sum():
    a=5
    b=4
    print(a+b)
sum()
```

```python
# function with arguments and no return type 
def sum(a,b):
    print(a+b)
sum(5,4)
```

```python
# function without arguments and  return type 
def sum():
    a=5
    b=4
    return(a+b)
a=sum()
print(a)
```

```python
# function with arguments and  return type
def sum(a,b):
    return(a+b)
a=sum(5,6)
print(a)
```
