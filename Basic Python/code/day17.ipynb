### Return multiple values

```python
def cal(a,b):
    add=a+b
    multi=a*b
    sub=a-b
    div=a/b
    return add,multi,sub,div
a,b,c,d=cal(10,5)
print(a)
print(b)
print(c)
print(d)
```

```python
# default arguments.
def sum(a,b=5):
    sum=a+b
    multi=a*b
    return sum,multi
summ=sum(5,19)
add,multi=summ
print(summ[0])
print(summ)
print(add)
print(multi)
```

```python
# wap using recursive to print from n to 1 
def print_reverse(n):
    if n==0:
         return
    else:
        print(n)
        print_reverse(n-1)

print_reverse(5)


     
     
```

```python
# wap using a recursive function to find the factorial of a number.
def find_factorial(n):
    if n==0 or n==1:
        return 1
    else:
        factorial=n*find_factorial(n-1)
        return factorial

fact=find_factorial(4)
print(fact)
    
```

```python
# write a recursive program to reverse a string 
text='hello my boy'
def reverse_string(text):
    if len(text)==0:
        return ' '
    
    else:
        return reverse_string(text[1:]) + text[0]
    
reverse_string(text)
```
