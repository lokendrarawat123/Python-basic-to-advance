```python
# wap a to create a biling system (product,price,qty,total,all total)
bill=''
n= int(input('Enter n = '))
for i in range(n):
    name=input('Enter name: ')
    price=int(input('Enter price:'))
    qty=int(input('enter quantity: '))
    total=price*qty
    data=f"{name} {price} {qty}\n"
    bill= bill + data

file=open('bill.txt','w')
file.write(bill)
file.close()
```

```python
file=open('bill.txt','r')
x=file.read()
print(x)
file.close()
```

```python
y=x.split("\n")[0:-1]
print(y)
```

```python
bill=[] 
for i in y:
    bill.append(i.split())
print(bill)
```
