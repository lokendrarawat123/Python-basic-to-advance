```python
from bs4 import BeautifulSoup,MarkupResemblesLocatorWarning
import warnings,requests
warnings.filterwarnings('ignore',category=MarkupResemblesLocatorWarning)
soup=BeautifulSoup('https://www.worldometers.info/coronavirus/','html.parser')
```

```python
data=requests.get(soup)
print(data)
```

```python
data=data.text
```

```python
row_data=BeautifulSoup(data,'lxml')
# row_data
```

```python
row_data.title.text
```

```python
row_data.title.name
```

```python
table_code=row_data.table

```

```python
tags=table_code.find_all('tr')
```

```python
data_rows=[]
for tag in tags:
    y=tag.text.split('\n')
    if y[1] !='':
        data_rows.append(y[1:])
# data_rows

    
```

```python
import csv
file=open('covid_data.csv','w',encoding='utf-8')
x=csv.writer(file)
x.writerows(data_rows)
file.close()
```
