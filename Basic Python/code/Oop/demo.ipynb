```python
import requests
from bs4 import BeautifulSoup
import pandas as pd

# 1. URL र Request
url = "https://www.worldometers.info/world-population/nepal-population/"
response = requests.get(url)

# 2. BeautifulSoup Parser
soup = BeautifulSoup(response.text, "html.parser")

# 3. Table र Rows समात्ने
table_code = soup.table
tags = table_code.find_all('tr')

data_rows = []

# 4. Simple For Loop प्रयोग गरेर Data निकाल्ने (List Comprehension नराखी)
for tag in tags:
    cells = tag.find_all(['th', 'td'])
    
    single_row = []
    for cell in cells:
        clean_text = cell.text.strip()
        single_row.append(clean_text)
    
    if single_row:
        data_rows.append(single_row)

# 5. Header र Data छुट्याउने
header_row = data_rows[0]    # पहिलो लाइन (Table Header/Column Names)
body_data = data_rows[1:]    # दोस्रो लाइनदेखि बाँकी (Actual Data)

# 6. Pandas DataFrame बनाउने
df = pd.DataFrame(body_data, columns=header_row)

# 7. CSV फाइलमा सेभ गर्ने
df.to_csv("nepal_population.csv", index=False,encoding='utf-8')

print("डाटा सफलतापूर्वक nepal_population.csv फाइलमा सेभ भयो!")
```
