```python

import matplotlib.pyplot as plt
import numpy as np
fig, ax = plt.subplots()             # Create a figure containing a single Axes.
ax.plot([1, 2, 3, 5], [1, 4, 2, 3])  # Plot some data on the Axes.
plt.show()                           # Show the figure.
```
