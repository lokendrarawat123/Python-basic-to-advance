### Polymorphism

Polymorphism is a core concept in Object-Oriented Programming (OOP) that means "many forms." It allows different classes to have methods with the exact same name, but each class can implement that method in its own unique way.

```python
class A:
    def method (self):
        pass
class B:
    def method(self):
        
        pass
obj=A()
obj1=B()
obj.method()
obj1.method()
```

```python
# example
class Parrot:
    def fly(self):
        print('Parrot can fly')
class Penguin:
    def fly(self):
        print('penguin cant fly')

blu=Parrot()
pe=Penguin()
blu.fly()
pe.fly()

```

```python
def bird_name(bird):
    bird.fly()
bird_name(blu)
bird_name(pe)
```

```python
class Shape:
    def __init__(self,shape):
        print (f'the shape is {shape}')
class Rectangle(Shape):
    def area(self):
        super().__init__(shape)
        print("this is ractanle")
class Circle(Shape):

    def area(self,shape):
        super().__init__(shape)
        print('this is circle')

a=Rectangle()
b=Circle()
a.area('rectangle')
b.area('circle')
    

```

```python
class Point:
    def __init__(self,x=0,y=0):
        self.x=x
        self.y=y
    # __add__(+),__sub__(-),__mul__(*),__turediv__(/),__floordiv__(//),__mod__(%),__pow__(**)
    def __add__(self,other):
        x=self.x+other.x
        y=self.y+other.y
        return Point(x,y)
    def __str__(self):
        return f"{self.x},{self.y}"

p1=Point(2,3)
p2=Point(2,3)
p3=Point(2,3)
p4=Point(2,3)
p5=Point(2,3)
p6=Point(2,3)
p7=Point(2,3)
p9=Point(2,3)
print(p1+p2+p3+p4+p5+p6+p7+p9)
```

```python
class A:
    def method(self):
        print(10)
class B(A):
    def method(self):
        super().method()
        print(100)

obj=B()
obj.method()  ## if same method then first run child class method 
```
