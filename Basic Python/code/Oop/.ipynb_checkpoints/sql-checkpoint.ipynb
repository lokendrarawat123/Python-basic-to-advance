```python
# wap a progra to nth number of students store data

n=int(input("enter a number: "))
for i in range(n):
    name=input(f'Enter {i+1}th student name: ')
    age=int(input(f'Enter {i+1}th student age: '))
    address=input(f'Enter {i+1}th student address: ')
    email=input(f'Enter {i+1}th student email: ')
    sql=f'''
    insert into students(name,age,address,email)
    values('{name}','{age}','{address}','{email}')
    '''
    cursor.execute(sql)
    conn.commit()
    print(f'data saved :{i+1}th data')

print("all data saved successfully")
```

```python
## take age in ascending order limit 5
sql='''
SELECT * FROM students ORDER BY age ASC 
'''
data=cursor.execute(sql)
data=cursor.fetchall()
for row in data:
    print(row)
print("data fetch succesfully")
```

```python
# select add where ktm and pokhara
sql='''
select address from students where address IN ('KTM','pokhara')
'''
data=cursor.execute(sql)
data=cursor.fetchall()
for row in data:
    print(row)
print("data fetch succesfully")
```

```python
sql='''
    select * from students where age BETWEEN 20 AND 24
'''
data=cursor.execute(sql)
data=cursor.fetchall()
for row in data:
    print(row)
print("data fetch succesfully")
```
