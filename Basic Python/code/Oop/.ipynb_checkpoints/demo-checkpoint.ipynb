```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
```

```python
url="https://www.worldometers.info/world-population/nepal-population/"
response=requests.get(url)
# response
response.raise_for_status()
```

```python
soup = BeautifulSoup(response.text,"html.parser")
```

```python
# soup
```

```python
table=soup.find('table',class_='datatable')
```

```python
# table
```

```python
headers_list=[th.text.strip() for th in table.find_all('th')]
```

```python
# headers_list
```

```python
# extracting data
data_rows=[]
for tr in table.find_all('tr')[1:]:
    cols=[td.text.strip().replace("\xa0","").replace("-","")for td in tr.find_all('td')]
    if cols:
        data_rows.append(cols)
```

```python
# data_rows
```

```python
df=pd.DataFrame(data_rows,columns=headers_list)
csv_filename='nepal_population_historical.csv'
df.to_csv(csv_filename,index=False,encoding="utf-8-sig")
print(f"Data succesfully scrapped and save to '{csv_filename}'!")
```

```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

# Load the scraped CSV file
df = pd.read_csv("nepal_population_historical.csv")

# Clean numeric data (remove commas and percentage signs)
df["Year"] = pd.to_numeric(df["Year"])
df["Population"] = df["Population"].str.replace(",", "").astype(int)
df["Yearly Change %"] = (
    df["Yearly % Change"].str.replace("%", "").astype(float)
)


df = df.sort_values("Year")

# Set aesthetic visual style
sns.set_theme(style="whitegrid")
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

# Plot 1: Total Population Growth
sns.lineplot(
    data=df,
    x="Year",
    y="Population",
    marker="o",
    ax=axes[0],
    color="#1f77b4",
    linewidth=2.5,
)
axes[0].set_title("Nepal Population Trend (1955–Present)", fontsize=14, pad=12)
axes[0].set_ylabel("Population (Millions)", fontsize=12)
axes[0].set_xlabel("Year", fontsize=12)
axes[0].yaxis.set_major_formatter("{x:,.0f}")


sns.barplot(
    data=df,
    x="Year",
    y="Yearly Change %",
    ax=axes[1],
    palette="Blues_d",
    hue="Year",
    legend=False,
)
axes[1].set_title(
    "Nepal Yearly Population Growth Rate (%)", fontsize=14, pad=12
)
axes[1].set_ylabel("Yearly Change (%)", fontsize=12)
axes[1].set_xlabel("Year", fontsize=12)
axes[1].tick_params(axis="x", rotation=45)

plt.tight_layout()
plt.show()
```
