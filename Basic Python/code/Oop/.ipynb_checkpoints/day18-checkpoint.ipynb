```python
class Hello:
    @staticmethod
    def hello(): 
        print('Hello')
    # @staticmethod
    def hello1():   # so we can use hello(self):
        print('HeLLo1')
obj=Hello()
obj.hello1()
```

```python
class Cal():
    def volume(self,l,b,h):
        print(l*b*h)
    def area(self,l,b):
        print(l*b)
obj=Cal()
obj.volume(10,10,10)
obj.area(10,10)
```

```python
class Cal:
    def __init__(self,l,b,h):
        self.l=l
        self.b=b
        self.h=h
    def volume(self):
        print(self.l*self.b*self.h)
    def area(self):
        print(self.l*self.b)
obj=Cal(10,10,10)
obj.volume()
obj.area()
```

```python
class Student:

    def __init__(self, name, marks):
        self.name = name
        # String list lai float ma convert garera numbers banako
        self.marks = [float(m) for m in marks]

    def show_details(self):
        print("\n" + "=" * 30)
        print("      STUDENT REPORT CARD")
        print("=" * 30)
        print(f" Name       : {self.name.title()}")
        print("-" * 30)

        # Harek subject ko mark formatting
        for i, mark in enumerate(self.marks, 1):
            print(f" Subject {i}  : {mark:.2f}")

        print("-" * 30)
        total = sum(self.marks)
        average = total / len(self.marks) if self.marks else 0

        print(f" Total Marks : {total:.2f}")
        print(f" Average     : {average:.2f}%")
        print("=" * 30)


# Input block
name = input("Enter Name: ")
marks_str = input("Enter marks (space-separated): ")

marks_list = marks_str.split()

# Object creation and method execution
obj = Student(name, marks_list)
obj.show_details()
```

```python
# ### create a class shoppingCart with methods :
# Conditions:
# getAmount()
# applyDiscout()
# displayBill()
# Apply 10% discount if amout exceeds Rs.5000.
```

```python
class ShoppingCart:

    def __init__(self, items):
        self.items = items
        self.total_amount = 0
        self.discount = 0
        self.final_amount = 0

    def getAmount(self):
        self.total_amount = sum(self.items.values())
        return self.total_amount

    def applyDiscount(self):
        self.getAmount()

        if self.total_amount > 5000:
            self.discount = self.total_amount * 0.10
        else:
            self.discount = 0

        self.final_amount = self.total_amount - self.discount
        return self.final_amount

    def displayBill(self):
        self.applyDiscount()

        print("--- BILL DETAILS ---")

        for item in self.items:
            print(item, ": Rs.", self.items[item])

        print("Total Amount:", self.total_amount)
        print("Discount:", self.discount)
        print("Final Bill Amount:", self.final_amount)


cart_items = {"Jacket": 3500, "Shoes": 2500, "Socks": 300}

obj = ShoppingCart(cart_items)
obj.displayBill()
```

```python
# ## create a class Electricitybill with methods :
# inputUnits()-calculateBill()-displayBill()
# Rules:
# first 100 untis=Rs.5/unit
# Next 100 units = rs.8/unit
# Remaining units=rs.10/unit
```

```python
class ElectricityBill:

    def input_units(self):
        units = float(input("Enter total units consumed: "))
        return units

    def calculate_bill(self):
        units = self.input_units()

        if units >= 0 and units <= 100:
            self.bill = units * 5

        elif units > 100 and units <= 200:
            self.bill = 500 + (units - 100) * 8

        elif units > 200:
            self.bill = 500 + 800 + (units - 200) * 10

        else:
            self.bill = 0
            print("Invalid unit entered!")

    def display_bill(self):
        self.calculate_bill()

        print(f"Total Electricity Bill = Rs. {self.bill:.2f}")
    

bill_obj = ElectricityBill()
bill_obj.display_bill()
```
