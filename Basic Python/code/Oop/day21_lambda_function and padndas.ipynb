### Lamda Function

```python
# lambda arg:expression
```

```python
def add(a):
    print(a%2==0)
add(10)
```

```python
y=lambda x:x%2==0
y(10)
```

```python
sum=lambda x,y:x+y
sum(5,8)
```

```python
even=lambda x:'even' if x%2==0 else 'odd'
print(even(12))
```

### Filter Function

```python
# list(filter(lambda arg:exp,data))
```

```python
l=[1,2,3,4,5,6,7]
y=list(filter(lambda x:2,l))
print(y)
```

```python
l=[1,2,3,4,5,6,7]
y=list(filter(lambda x:x%2==0,l))
print(y)
```

```python
## map
# adding each by 3
```

```python
l=[1,2,3,4,5,6,7]
y=list(map(lambda x:x+3,l))
print(y)
```

```python
# wap to apply a 15 % discount on a list of product prices using the map() function and lambda function.
list1=[500,1999,456,445,600,500,200]
discount=list(map(lambda x: x-x*0.15,list1))
print(f' with 15% discount list : {discount}') 
```

```python
##  wap to convert all words in a list into uppercase using map() function and lambda function

list2=['lokendra','binaya','rajkumar','hello','bishal']
uppercase=list(map(lambda name: name.upper(),list2))
print(uppercase)
```

```python
## Reduce
# reduce(function ,operation)
from functools import reduce
lis=[1,2,3,4,5,6,7]
data=reduce(lambda a,b:a+b,lis)
print(data)
```

```python
# largest
list3=[1,2,3,4,5,6,7]
largest=reduce(lambda a,b:a if a>b else b,list3)
print(largest)
```

```python
list3=[1,2,3,4,5,6,7]
largest=reduce(lambda a,b:a if a>b else b,list3)
second_largest=reduce(lambda a,b:a if a>b else b,[x for x in list3 if x!=largest])
print(second_largest)
```
