```python
# wap to input `n` peoples names and phone numbers and store them in a dictionary.finally display the dictionary.
a={}
n=int(input('Enter the number of people'))
for i in range(1,n+1):
    name=input(f'Enter the name of {i}th people: ')
    phone=int(input(f'enter the phone number: '))
    a.update({name:phone})
print(a)
for i in a.items():
    print(i)
    
```

```python
#  wap a program to store multiple users name age ,and address in a dictionary and display the data.
# # a={
#         'name':['ram','hari','shyam']
#         'age':[24,54,54],
#         'address':['kathmandu','bhaktapur','lalitpur']


a={
    'name':[],
    'age':[],
    'address':[],
}
n=int(input('Enter the number of people'))
for i in range(1,n+1):
    name=input(f'Enter the name of {i}th people: ')
    age=int(input(f'enter the age  {i}th people: '))
    address=input(f'enter the address of {i}th people')
    a['name'].append(name)
    a['age'].append(age)
    a['age'].append(address)
print(a)
    
    
```
