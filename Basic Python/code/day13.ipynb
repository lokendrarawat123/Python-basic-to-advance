```python
a = [[1, 2, 3, 4], [2, 3, 4, 5], [4, 3, 2, 1]]
b = [
    [1, 2, 3, 4, 5],
    [3, 2, 3, 1, 3],
    [4, 4, 3, 3, 2],
    [2, 3, 4, 2, 1],
]

c = []  # Empty list baat start garne

for i in range(len(a)):
    row = []  # Har euta A-row ko laghi nayan row create garne

    for j in range(len(b[0])):
        total = 0  # Product dynamic sum calculate garne variable

        for k in range(len(b)):
            total += a[i][k] * b[k][j]

        row.append(total)  # Cell ko total sum row list ma append garne

    c.append(row)  # Ready bhaeko complete row main matrix 'c' ma append garne

print(c)
```
