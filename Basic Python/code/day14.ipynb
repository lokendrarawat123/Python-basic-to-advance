### Tuple


```python
# tuple : indxing ,multiple dublicate data,Immutable
a=tuple()
print(type(a))
# we cant modify or change the value of tuple and if we need to change then we convert in list then perform task.
```

```python
a=(1,2,3,4,5,6)
# for change tuple to list need type cast 
list1=list(a)
print(list1)
print(max(a))
print(min(a))
print(sum(a))
```

```python
# List comprehension
a=[1,2,3,4,5,6]
b=[i for i in a if i%2==0]
print(b)
```

```python
nums=[1,4,6,8,3,9]
b=[i for i in nums if i>=5]
print(b)
```

```python
words=['cat','dog','apple','sun','ball']
a=[ i for i in words  if 'a'  in i] 
print(a)
```

```python
# chnage tuple to list
a=(('ram',78,'kathmandu'),('shyam',78,'jhapa'))
list1=[]
for i in a :
    list1.append(list(i))

print(list1)
        
        
```

```python
a=(('ram',78,'kathmandu'),('shyam',78,'jhapa'))
list1=[list(i) for i in a]


print(list1)
        
```

```python
a=[['ram', 78, 'kathmandu'], ['shyam', 78, 'jhapa']]
list2=(tuple(i) for i in a)
print(list2)
```
