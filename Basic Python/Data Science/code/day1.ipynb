## Population(N)

-A Population is the complete set off all individuals or data values you want to study.
    -It includes every element of the group.
    -Population charactersistics(mean,standard deviation,proportion,median,percentiles,etc) are foxed values   because they describe the entire group.
    -These overall charactersitics are known as the population distribution.

## Sample(n)

-A sample is asmaller subset taken from the population.
    -It includes only some values that you actually use for analysis or estiomation.
    -Fromt this sample ,we calculate quantities like sample mean,sample standart devitaion and sample population to estimate the population values.
    -These calculated values describe the sample distribution

### Descriptive Statistics

-This method offers a way to summarize data by transforming raw observation into meaningfull information that is easy to interpret and share


```python
Types:
Measures of Central Tendency: Identify the center or average value of a dataset (Mean, Median, Mode).

Measures of Dispersion / Variability: Show how spread out the data points are (Range, Variance, Standard Deviation, Interquartile Range).

Percentile: Track how often specific values or categories occur (Counts, Percentages, Frequency Distributions).


Measures of Shape: Describe the symmetry and peakness of the data distribution (Skewness, Kurtosis).
Measures of Position: Describe the location of a specific data point relative to the whole set (Percentiles, Quartiles, Z-scores).
Graphical / Visual Descriptions: Present data trends visually using structured diagrams (Histograms, Box Plots, Bar Charts, Scatter Plots).
```


## Mean

```python
import pandas as pd
import numpy as np
arr=np.array([1,2.456,3,4,5,6.678,7,88,9])
length=len(arr)
sum=np.sum(arr)
mean=sum/length
print(f'Mean of the array is :{mean}')
```
